```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import OneHotEncoder
from sklearn.metrics.pairwise import cosine_similarity

print("Libraries imported successfully!")
```

```python

# Define the features used by our recommendation model

feature_columns = [
    "family",
    "long_stay",
    "business",
    "airport_arrival",
    "spa_interest",
    "breakfast_need",
    "standard_room"
]

# Create service profiles

service_features = pd.DataFrame({

    "service_name": [
        "Breakfast Package",
        "Airport Pickup",
        "Spa Package",
        "Late Checkout",
        "Room Upgrade",
        "Laundry Service",
        "Family Package"
    ],

    "family": [1, 0, 0, 0, 0, 0, 1],
    "long_stay": [1, 0, 0, 1, 0, 1, 1],
    "business": [0, 1, 0, 1, 1, 0, 0],
    "airport_arrival": [0, 1, 0, 0, 0, 0, 0],
    "spa_interest": [0, 0, 1, 0, 0, 0, 0],
    "breakfast_need": [1, 0, 0, 0, 0, 0, 1],
    "standard_room": [0, 0, 0, 0, 1, 0, 0]

})

# Display service features

service_features
```

```python

# Create a sample guest profile

sample_guest = pd.DataFrame([{
    "family": 1,
    "long_stay": 1,
    "business": 0,
    "airport_arrival": 0,
    "spa_interest": 0,
    "breakfast_need": 1,
    "standard_room": 1
}])

sample_guest
```

```python

from sklearn.metrics.pairwise import cosine_similarity

# Extract numerical service features

service_matrix = service_features[feature_columns]

# Calculate similarity between guest and services

similarity_scores = cosine_similarity(
    sample_guest[feature_columns],
    service_matrix
)

# Create a results table

recommendation_results = pd.DataFrame({

    "service_name": service_features["service_name"],

    "similarity_score": similarity_scores[0]

})

# Sort from highest to lowest similarity

recommendation_results = recommendation_results.sort_values(
    by="similarity_score",
    ascending=False
)

recommendation_results
```

```python

# Select the top 3 recommendations

top_recommendations = recommendation_results.head(3)

print("TOP 3 PERSONALIZED RECOMMENDATIONS")
print("----------------------------------")

for index, row in top_recommendations.iterrows():

    print(
        f"{row['service_name']}: "
        f"{row['similarity_score']:.2f}"
    )
```

```python

# Create another guest profile

business_guest = pd.DataFrame([{
    "family": 0,
    "long_stay": 0,
    "business": 1,
    "airport_arrival": 1,
    "spa_interest": 0,
    "breakfast_need": 0,
    "standard_room": 1
}])

# Calculate similarity

business_scores = cosine_similarity(
    business_guest[feature_columns],
    service_matrix
)

# Rank services

business_results = pd.DataFrame({

    "service_name": service_features["service_name"],

    "similarity_score": business_scores[0]

}).sort_values(
    by="similarity_score",
    ascending=False
)

print("BUSINESS GUEST RECOMMENDATIONS")

business_results.head(3)
```

```python

def recommend_for_guest(guest_profile, top_n=3):

    # Calculate similarity between guest and services
    scores = cosine_similarity(
        guest_profile[feature_columns],
        service_features[feature_columns]
    )[0]

    # Store service names and similarity scores
    results = pd.DataFrame({
        "service_name": service_features["service_name"],
        "similarity_score": scores
    })

    # Remove services with zero similarity
    results = results[results["similarity_score"] > 0]

    # Sort by highest similarity
    results = results.sort_values(
        by="similarity_score",
        ascending=False
    )

    # Return the top recommendations
    return results.head(top_n).reset_index(drop=True)
```

```python

family_recommendations = recommend_for_guest(sample_guest)

print("FAMILY GUEST RECOMMENDATIONS")
display(family_recommendations)
```

```python

business_recommendations = recommend_for_guest(business_guest)

print("BUSINESS GUEST RECOMMENDATIONS")
display(business_recommendations)
```

```python

def explain_recommendation(service_name, guest):

    explanations = {
        "Breakfast Package": "This guest may benefit from an additional breakfast option.",
        "Airport Pickup": "Airport pickup may provide convenient transportation for the guest.",
        "Spa Package": "This service matches the guest's recorded wellness interest.",
        "Late Checkout": "Late checkout may provide additional flexibility.",
        "Room Upgrade": "The guest's room profile suggests an upgrade may be relevant.",
        "Laundry Service": "This service may be useful for guests staying longer.",
        "Family Package": "This package may suit guests travelling with children."
    }

    return explanations.get(
        service_name,
        "This service matches some features of the guest profile."
    )
```

```python

def guestly_recommendations(guest_profile, top_n=3):

    recommendations = recommend_for_guest(
        guest_profile,
        top_n
    )

    recommendations["reason"] = recommendations["service_name"].apply(
        lambda service: explain_recommendation(
            service,
            guest_profile
        )
    )

    return recommendations
```

```python
family_output = guestly_recommendations(sample_guest)

display(family_output)
```

```python
business_output = guestly_recommendations(business_guest)

display(business_output)
```
